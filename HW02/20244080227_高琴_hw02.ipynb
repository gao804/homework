## 一、字符串的乐趣

### 1. is_palindrome

先去掉空格再统一转小写，然后和倒过来的字符串比较。

```python
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]


for s in ["level", "kayak", "tacocat", "T A C O cat", "rats live on no evil star", "python"]:
    print(repr(s), is_palindrome(s))
```

### 2. is_abecedarian

去掉空格后，只要每个字母都不比后一个字母大，就是字母顺序的。相同字母相邻也可以。

```python
def is_abecedarian(s):
    s = s.replace(" ", "").lower()
    for i in range(len(s) - 1):
        if s[i] > s[i + 1]:
            return False
    return True


for s in ["adder", "beet", "abcd efgh xyz", "dog", "cat"]:
    print(repr(s), is_abecedarian(s))
```

### 3. remove_vowels

```python
def remove_vowels(s):
    return "".join(c for c in s if c.lower() not in "aeiou")


print(repr(remove_vowels("cat")))
print(repr(remove_vowels("audio")))
print(repr(remove_vowels("aeiou")))
print(repr(remove_vowels("Hello World")))
```

## 二、列表的乐趣

### 1. list_reverse

不是列表时抛出 TypeError。用切片返回新列表，原列表不变。

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError(f"需要列表，传入的是 {type(lst).__name__}")
    return lst[::-1]


a = [1, 2, 3]
print(list_reverse(a), a)
print(list_reverse([]))
try:
    list_reverse((1, 2, 3))
except TypeError as e:
    print("TypeError:", e)
```

### 2. is_sorted

从头到尾只走一遍，碰到前一个比后一个大就直接返回 False。函数里只用了下标和比较，所以列表、元组、字符串都能用。

```python
def is_sorted(seq):
    for i in range(1, len(seq)):
        if seq[i - 1] > seq[i]:
            return False
    return True


print(is_sorted([1, 2, 2, 5]))
print(is_sorted([3, 1, 2]))
print(is_sorted((1.5, 2, 7)))
print(is_sorted("abcz"))
print(is_sorted("hello"))
print(is_sorted([]))
```

## 三、作为向量的元组

### 1. vec_scalar_mult

```python
def is_number(x):
    return isinstance(x, (int, float)) and not isinstance(x, bool)


def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组")
    if not all(is_number(x) for x in t):
        raise TypeError("向量的元素必须是整数或浮点数")
    if not is_number(s):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(x * s for x in t)


print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((1.5, -2), 0.5))
for args in [([1, 2], 3), ((1, "2"), 3), ((1, 2), "3")]:
    try:
        vec_scalar_mult(*args)
    except TypeError as e:
        print("TypeError:", e)
```

### 2. vec_inner_product

```python
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个向量都必须是元组")
    if len(x) != len(y):
        raise ValueError(f"长度不一致：{len(x)} 和 {len(y)}")
    total = 0.0
    for a, b in zip(x, y):
        total += a * b
    return total


print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((), ()))
try:
    vec_inner_product((1, 2), [3, 4])
except TypeError as e:
    print("TypeError:", e)
try:
    vec_inner_product((1, 2), (3, 4, 5))
except ValueError as e:
    print("ValueError:", e)
```

### 3. check_valid_mx

空元组不当作矩阵，返回 False。

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple) or len(mx) == 0:
        return False
    if not all(isinstance(row, tuple) for row in mx):
        return False
    n = len(mx[0])
    for row in mx:
        if len(row) != n:
            return False
        if not all(is_number(x) for x in row):
            return False
    return True


print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2, 3),)))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, 2), [3, 4])))
print(check_valid_mx(((1, "a"), (3, 4))))
print(check_valid_mx([[1, 2], [3, 4]]))
```

### 4. mx_vec_mult

m×n 矩阵乘长度为 n 的向量，结果的第 i 个数就是第 i 行和向量的内积。
矩阵不合法时报 TypeError，长度对不上时报 ValueError。

```python
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise TypeError("矩阵必须是由等长数值元组组成的元组")
    if not isinstance(v, tuple) or not all(is_number(x) for x in v):
        raise TypeError("向量必须是由数值组成的元组")
    if len(v) != len(mx[0]):
        raise ValueError(f"矩阵有 {len(mx[0])} 列，向量长度是 {len(v)}")
    return tuple(vec_inner_product(row, v) for row in mx)


A = ((1, 2, 3), (4, 5, 6))
print(mx_vec_mult(A, (1, 0, -1)))
print(mx_vec_mult(((2, 0), (0, 2)), (3, 4)))
for args in [(((1, 2), (3,)), (1, 1)), (A, (1, 2))]:
    try:
        mx_vec_mult(*args)
    except (TypeError, ValueError) as e:
        print(type(e).__name__ + ":", e)
```

## 四、更多向量的乐趣

### 1~5. Vector 类

只给维度时，元素默认全是 0。比较运算写了 `__eq__` 和 `__lt__`，其余几个交给 `functools.total_ordering` 生成。
元素存成元组，比较时直接用元组的字典序。

```python
from functools import total_ordering


@total_ordering
class Vector:
    def __init__(self, dim, elements=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能是负数")
        if elements is None:
            elements = (0,) * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if not all(is_number(x) for x in elements):
            raise TypeError("元素必须是整数或浮点数")
        if len(elements) != dim:
            raise ValueError(f"维度是 {dim}，元素却有 {len(elements)} 个")
        self.dim = dim
        self.elements = tuple(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    def _check_other(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和 Vector 对象运算")
        if self.dim != other.dim:
            raise ValueError(f"维度不同：{self.dim} 和 {other.dim}")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_other(other)
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        self._check_other(other)
        return self.elements < other.elements

    def dot(self, other):
        self._check_other(other)
        return sum(a * b for a, b in zip(self.elements, other.elements))

    def __mul__(self, other):
        if isinstance(other, Vector):
            self._check_other(other)
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        if is_number(other):
            return Vector(self.dim, [a * other for a in self.elements])
        return NotImplemented

    __rmul__ = __mul__
```

#### 初始化

```python
print(Vector(3, [1, 2, 3]))
print(Vector(2, (1.5, -2)))
print(Vector(4))
print(Vector(0))
for args in [(2.0, (1, 2)), (-1, ()), (3, (1, 2)), (2, (1, "x")), (2, "ab")]:
    try:
        Vector(*args)
    except (TypeError, ValueError) as e:
        print(f"Vector{args}", type(e).__name__ + ":", e)
```

#### 2. 维度和元素应该是实例属性

每个向量都有自己的维度和元素，`Vector(2, (2, 4))` 和 `Vector(3, (1, 2, 3))` 不一样。
如果放成类属性，所有对象共用同一份数据，改一个就全变了，所以在 `__init__` 里用 `self.dim`、`self.elements` 保存。
下面两个向量各自保存了不同的值，类本身上没有这两个属性。

```python
u = Vector(2, (2, 4))
w = Vector(3, (1, 2, 3))
print(u.dim, u.elements)
print(w.dim, w.elements)
print(hasattr(Vector, "dim"), hasattr(Vector, "elements"))
```

#### 3. 比较运算

```python
a = Vector(2, (2, 4))
b = Vector(2, (2, 5))
c = Vector(2, [2, 4])
print("a == c:", a == c)
print("a != b:", a != b)
print("a < b:", a < b)
print("a <= c:", a <= c)
print("b > a:", b > a)
print("b >= a:", b >= a)
print("a == (2, 4):", a == (2, 4))
try:
    a < Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

#### 4. dot

```python
v = Vector(3, (1, 2, 3))
w = Vector(3, (4, 5, 6))
print(v.dot(w))
for other in [(4, 5, 6), Vector(2, (1, 1))]:
    try:
        v.dot(other)
    except (TypeError, ValueError) as e:
        print(type(e).__name__ + ":", e)
```

#### 5. 乘法

```python
v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print("2 * v =", 2 * v)
print("v * 2 =", v * 2)
print("v * 0.5 =", v * 0.5)
print("v * w =", v * w)
print(v)
try:
    v * Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
try:
    v * None
except TypeError as e:
    print("TypeError:", e)
```
