## 实验2：函数、条件、循环与递归

### 2.1 成绩分级
先排除范围外的成绩，再从高分到低分判断。例如 85 先不满足 90 分条件，随后满足 80 分条件，返回“良好”。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in (0, 59, 60, 79, 80, 85, 89, 90, 100):
    print(score, score_level(score))
for score in (-1, 101):
    try:
        score_level(score)
    except ValueError as error:
        print(f"{score} -> ValueError: {error}")
```

### 2.2 用 for 和 while 求和
两种方法都把 1 到 100 逐个加到累计值中。`while` 需要手动更新计数器，否则循环不会结束。

```python
total_for = 0
for number in range(1, 101):
    total_for += number

total_while = 0
number = 1
while number <= 100:
    total_while += number
    number += 1

print("for 求和：", total_for)
print("while 求和：", total_while)
print("两者相等：", total_for == total_while)
assert total_for == total_while == 5050
```

### 2.3 递归与迭代阶乘
递归把 `5!` 分成 `5 * 4!`，一直到 `0! = 1` 后逐层返回。
迭代从 1 开始，依次乘以 2、3、4、5。两者结果相同，迭代不需要额外的递归调用层数。

```python
def factorial_recursive(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for factor in range(2, n + 1):
        result *= factor
    return result


for n in (0, 1, 5, 10):
    recursive = factorial_recursive(n)
    iterative = factorial_iterative(n)
    print(f"{n}!：递归={recursive}，迭代={iterative}，相等={recursive == iterative}")
    assert recursive == iterative
```

### 2.4 返回元组并解包
函数一次返回三个结果；调用者按相同顺序分别接收最小值、最大值和平均值。

```python
def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


minimum, maximum, average = summary([85, 58, 92, 76, 60, 89])
print(f"最小值={minimum}，最大值={maximum}，平均值={average:.2f}")
```

### 2.5 可变数量参数
`*args` 将传入的位置参数收集成元组；没有参数时，空元组的和是 0。

```python
def sum_all(*args):
    return sum(args)


print("无参数：", sum_all())
print("单个参数：", sum_all(7))
print("多个参数：", sum_all(1, 2, 3, 4))
assert (sum_all(), sum_all(7), sum_all(1, 2, 3, 4)) == (0, 7, 10)
```

## 实验4：文件读写、异常处理与模块

### 4.1-4.2 创建目录、写入并读取 UTF-8 文件
数据流：题目文本 → `exp01_data/students.txt` → 跳过表头后的数据行。
路径相对于 Notebook 所在目录构造，不依赖固定盘符。每次运行先用 `w` 重建输入文件。

```python
print("当前工作目录：", os.getcwd())
data_dir = "exp01_data"
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")
students_text = "\n".join([
    "name,score", "Alice,85", "Bob,58", "Carol,92", "David,76",
    "Eva,60", "Frank,89", "Grace,abc", "Henry,105", "Incomplete",
]) + "\n"
with open(students_path, "w", encoding="utf-8") as file:
    file.write(students_text)
with open(students_path, "r", encoding="utf-8") as file:
    header = next(file).strip()
    data_lines = list(file)
print("表头：", header)
print("读取的数据行数：", len(data_lines))
print("".join(data_lines), end="")
```

### 4.3 解析一条学生记录
例如 `Alice,85` 先按逗号拆成两个字段，再把成绩转为整数，最后得到 `{"name": "Alice", "score": 85}`。
字段数量、空姓名、整数格式或成绩范围不正确时，用 `ValueError` 告诉调用者该行无效。

```python
def parse_student(line):
    """把一行 name,score 转为字典；无效记录抛出 ValueError。"""
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量必须为 2（姓名,成绩）")
    name, score_text = (field.strip() for field in fields)
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(score_text)
    except ValueError as error:
        raise ValueError("成绩必须是整数") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}
```

### 4.4 逐行处理并保留错误原因
每行单独捕获异常，所以一条坏记录不会终止整批读取。行号从 2 开始，因为第 1 行是表头。
有效记录放进字典列表，无效记录保留原始行号和原因。

```python
valid_records = []
invalid_records = []
for line_number, line in enumerate(data_lines, start=2):
    try:
        valid_records.append(parse_student(line))
    except ValueError as error:
        invalid_records.append({"line": line_number, "reason": str(error)})
        print(f"第 {line_number} 行无效：{error}")
print("有效记录：", valid_records)
print(f"有效 {len(valid_records)} 条，无效 {len(invalid_records)} 条")
```

### 4.5 输出报告，比较 w 和 a
`w` 打开文件会清空旧内容；`a` 保留旧内容并在末尾追加。
这里先写一行演示旧内容，再用 `w` 写统计结果，最后用 `a` 追加错误明细。
因此重复运行本单元时不会重复累积上一次的报告。

```python
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as file:
    file.write("这条旧内容将被覆盖。\n")
with open(report_path, "r", encoding="utf-8") as file:
    print("覆盖前：", file.read().strip())

valid_scores = [record["score"] for record in valid_records]
report_mean = sum(valid_scores) / len(valid_scores)
report_pass_rate = sum(score >= 60 for score in valid_scores) / len(valid_scores)
with open(report_path, "w", encoding="utf-8") as file:
    file.write(f"有效记录数：{len(valid_records)}\n")
    file.write(f"无效记录数：{len(invalid_records)}\n")
    file.write(f"平均分：{report_mean:.2f}\n")
    file.write(f"及格率：{report_pass_rate:.2%}\n")
with open(report_path, "r", encoding="utf-8") as file:
    overwritten_text = file.read()
print("w 模式覆盖后：\n" + overwritten_text)
assert "这条旧内容" not in overwritten_text

with open(report_path, "a", encoding="utf-8") as file:
    file.write("错误明细（a 模式追加）：\n")
    for record in invalid_records:
        file.write(f"第 {record['line']} 行：{record['reason']}\n")
with open(report_path, "r", encoding="utf-8") as file:
    final_report = file.read()
print("a 模式追加后：\n" + final_report)
assert final_report.startswith(overwritten_text)
```

`pickle` 可以保存 Python 的列表和字典结构。这里只读取本单元刚生成的文件，并比较恢复后的数据是否相等。

## 实验5：类、继承与对象复制

### 5.1-5.2 Student 类与实例
一个 `Student` 对象把姓名、成绩和判断是否及格的方法放在一起。
从实验4的有效记录创建对象；`print(student)` 会调用 `__str__` 得到可读的文字。

```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


students = [Student(record["name"], record["score"]) for record in valid_records]
for student in students:
    print(f"{student}，是否及格：{student.is_passed()}")
```

### 5.3 GraduateStudent 继承
研究生仍然有姓名、成绩及及格判断，因此通过继承复用这些行为。
`super().__init__()` 初始化父类属性，再补充研究方向；重写 `__str__` 让输出包含研究方向。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return (f"GraduateStudent(name={self.name}, score={self.score}, "
                f"research_topic={self.research_topic})")


graduate = GraduateStudent("Alice", 85, "数据分析")
print(graduate)
print("继承的及格判断：", graduate.is_passed())
```

### 5.4 直接赋值、浅复制与深复制
每次从新的 `{"name": "Alice", "scores": [85, 90]}` 开始。
直接赋值共用整个字典；浅复制创建外层字典，但内部成绩列表仍共用；深复制连内部列表也复制。
所以修改复制对象的内部列表时，前两种会影响原对象，深复制不会。

```python
import copy

for mode in ("直接赋值", "浅复制", "深复制"):
    original = {"name": "Alice", "scores": [85, 90]}
    if mode == "直接赋值":
        duplicated = original
    elif mode == "浅复制":
        duplicated = copy.copy(original)
    else:
        duplicated = copy.deepcopy(original)
    duplicated["scores"][0] = 100
    print(mode)
    print("  原对象：", original)
    print("  复制对象：", duplicated)
    print("  外层对象相同：", duplicated is original)
    print("  内部列表相同：", duplicated["scores"] is original["scores"])
```
